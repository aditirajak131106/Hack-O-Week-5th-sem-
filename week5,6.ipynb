```python
import numpy as np
import matplotlib.pyplot as plt
```

```python
#vector graph
donor1 = np.array([25, 3])
donor2 = np.array([30, 5])

plt.figure(figsize=(7, 5))

plt.quiver(0, 0, donor1[0], donor1[1],
           angles='xy', scale_units='xy', scale=1,
           label="Donor 1")

plt.quiver(0, 0, donor2[0], donor2[1],
           angles='xy', scale_units='xy', scale=1,
           label="Donor 2")

plt.xlim(0, 35)
plt.ylim(0, 8)
plt.xlabel("Age")
plt.ylabel("Number of Donations")
plt.title("Donor Vectors")
plt.grid()
plt.legend()

plt.show()
```

```python
#dot product
donors = np.array([
    [25, 3],
    [30, 5],
    [22, 2],
    [35, 6],
    [28, 4]
])

weights = np.array([0.2, 0.5])

scores = donors @ weights

names = ["Amit", "Priya", "Rahul", "Sneha", "Karan"]

plt.figure(figsize=(8, 5))

plt.bar(names, scores)

plt.xlabel("Donors")
plt.ylabel("Donor Score")
plt.title("Blood Donor Weighted Scores")

plt.show()
```

```python
#matrixdata graph
donor_matrix = np.array([
    [25, 3],
    [30, 5],
    [22, 2],
    [35, 6],
    [28, 4]
])

plt.figure(figsize=(7, 5))

plt.imshow(donor_matrix, cmap="Blues")

plt.colorbar(label="Value")

plt.xticks([0, 1], ["Age", "Donations"])
plt.yticks(range(5), names)

plt.title("Donor Data Matrix")

plt.show()
```

```python
#eigen value graph
matrix = np.array([
    [4, 1],
    [2, 3]
])

eigenvalues, eigenvectors = np.linalg.eig(matrix)

plt.figure(figsize=(7, 5))

plt.bar(
    ["Eigenvalue 1", "Eigenvalue 2"],
    eigenvalues
)

plt.xlabel("Eigenvalues")
plt.ylabel("Value")
plt.title("Eigenvalues of Donor Data Matrix")

plt.show()

print("Eigenvalues:", eigenvalues)
```

```python
#eigenvector graph
origin = np.zeros(2)

plt.figure(figsize=(7, 7))

for i in range(2):
    vector = eigenvectors[:, i]

    plt.quiver(
        origin[0],
        origin[1],
        vector[0],
        vector[1],
        angles='xy',
        scale_units='xy',
        scale=1,
        label=f"Eigenvector {i+1}"
    )

plt.xlim(-1, 1)
plt.ylim(-1, 1)

plt.xlabel("X")
plt.ylabel("Y")
plt.title("Eigenvectors")

plt.grid()
plt.legend()

plt.show()
```
