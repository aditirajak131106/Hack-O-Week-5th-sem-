```python
#Import libraries
#Imports the NumPy, Pandas, Matplotlib, and Seaborn libraries required for data analysis and visualization.
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
#blood donar finder dataset
#Creates a Blood Donor Finder dataset and stores it in a Pandas DataFrame.
blood_data = {
    "Donor_ID":[101,102,103,104,105,106,107,108],
    "Name":["Amit","Priya","Rahul","Sneha","Karan","Neha","Rohan","Pooja"],
    "Blood_Group":["A+","B+","O+","AB+","A-","O+","B-","A+"],
    "Age":[22,25,28,30,35,27,40,24],
    "City":["Nagpur","Mumbai","Nagpur","Pune","Delhi","Nagpur","Mumbai","Pune"],
    "Units_Donated":[2,3,4,2,5,1,4,3],
    "Availability":["Yes","No","Yes","Yes","No","Yes","Yes","No"]
}

df = pd.DataFrame(blood_data)
df
```

```python
# Function to count available donors
#Defines a function to display all available blood donors from the dataset.
def available_donors(data):
    return data[data["Availability"]=="Yes"]

    print("Available Donors")
    available_donors(df)
```

```python
# Search donors by blood group
#Defines a function to search donors based on a selected blood group.
def search_blood_group(group):
    return df[df["Blood_Group"]==group]

    search_blood_group("O+")
```

```python
#OOP
#Demonstrates Object-Oriented Programming by creating a Donor class and displaying donor details.
class Donor:

    def __init__(self,name,blood,city):
        self.name=name
        self.blood=blood
        self.city=city

    def display(self):
        print("Donor Name :",self.name)
        print("Blood Group :",self.blood)
        print("City :",self.city)

d1=Donor("Amit","A+","Nagpur")
d1.display()
```

```python
#list comprehension Donors above age 25
#Uses list comprehension to create a list of donors whose age is greater than 25 years.
age_list=[age for age in df["Age"] if age>25]

print(age_list)
```

```python
#dictionary comprehension
#Uses dictionary comprehension to create a dictionary mapping donor names to their blood groups.
donor_dict={row["Name"]:row["Blood_Group"] for index,row in df.iterrows()}

donor_dict
```

```python
#NumPy
#Converts the donor age and blood donation units into NumPy arrays for numerical processing.
ages=np.array(df["Age"])

units=np.array(df["Units_Donated"])

print(ages)
print(units)
```

```python
#Broadcasting
#Assume every donor donates one extra unit
#Demonstrates NumPy broadcasting by increasing every donor's donated units by one.
new_units=units+1

print(new_units)
```

```python
#vectorized Operation
#Performs vectorized operations to calculate statistical values such as mean, maximum, minimum, and square root of ages.
print("Average Age :",np.mean(ages))

print("Maximum Units :",np.max(units))

print("Minimum Units :",np.min(units))

print("Square Root of Age")

print(np.sqrt(ages))
```

```python
#panda
#data frame
#Displays the first few rows of the Blood Donor Finder dataset.
df.head()
```

```python
# data cleaning
#Add one missing value
#Cleans the dataset by replacing missing values with the average donated units.
df.loc[2,"Units_Donated"]=None

print(df)

df["Units_Donated"]=df["Units_Donated"].fillna(df["Units_Donated"].mean())

df
```

```python
#merge
#Merges the donor dataset with hospital information based on the city.
hospital=pd.DataFrame({
    "City":["Nagpur","Mumbai","Pune","Delhi"],
    "Hospital":["City Hospital","Apollo","Ruby Hall","AIIMS"]
})

merged=pd.merge(df,hospital,on="City")

merged
```

```python
#groupBy
#Groups donors by blood group and calculates the total blood units donated for each group.
group=df.groupby("Blood_Group")["Units_Donated"].sum()

print(group)
```

```python
#matplotlib
#blood group distribution
#Creates a Matplotlib bar chart showing blood units donated by each blood group.
plt.figure(figsize=(8,5))

plt.bar(df["Blood_Group"],df["Units_Donated"])

plt.title("Units Donated by Blood Group")

plt.xlabel("Blood Group")

plt.ylabel("Units")

plt.show()
```

```python
#Age histogram
#Creates a histogram to visualize the age distribution of blood donors.
plt.figure(figsize=(8,5))

plt.hist(df["Age"],bins=5)

plt.title("Age Distribution")

plt.xlabel("Age")

plt.ylabel("Frequency")

plt.show()
```

```python
#seaborn
#count plot
#Uses Seaborn to display the number of donors in each blood group.
plt.figure(figsize=(8,5))

sns.countplot(data=df,
              x="Blood_Group")

plt.show()
```

```python
#scatter plot
#Creates a scatter plot showing the relationship between donor age and blood units donated.
plt.figure(figsize=(8,5))

sns.scatterplot(data=df,
                x="Age",
                y="Units_Donated",
                hue="Blood_Group",
                s=120)

plt.show()
```

```python
#Box plot
#Creates a box plot to compare the distribution of blood donation units across different blood groups.
plt.figure(figsize=(8,5))

sns.boxplot(data=df,
            x="Blood_Group",
            y="Units_Donated")

plt.show()
```

```python
#heatmap
#Generates a heatmap to visualize the correlation between donor age and blood units donated.
plt.figure(figsize=(5,4))

sns.heatmap(df[["Age","Units_Donated"]].corr(),
            annot=True,
            cmap="coolwarm")

plt.show()
```
